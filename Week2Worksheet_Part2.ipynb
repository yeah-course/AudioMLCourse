## 1 - Work with Data using Pandas

We’ll start by generating a small synthetic dataset.  
Later, you’ll load a dataset from a CSV file for your challenge.

```python
import pandas as pd # Pandas is a data analysis and manipulation package
import numpy as np

# Create synthetic dataset
np.random.seed(42)
x = np.linspace(0, 10, 50) # Create our input variable, x
y = 2.5 * x + 1.0 + np.random.normal(0, 2, size=len(x)) # Create our target variable, y

# Put into a pandas DataFrame
df = pd.DataFrame({"Input Variable": x, "Target Variable": y})
df.head()
```

The Pandas DataFrame is a convinient Python object for working with tabular data.
`df.head()` displays the first five entries in the Dataframe.

### 1.1 Splitting Features and Target

We’ll separate the input (`input variable`) from the target (`target variable`).  
Scikit-learn expects the input variables (or, *features*) to be in **2D array** form.

```python
x_values = df[["Input Variable"]]   # double brackets -> keeps 2D shape
y_values = df["Target Variable"]     # single bracket -> 1D Series
```

## 2 - Fitting a Linear Regression Model

We’ll use scikit-learn’s `LinearRegression` model.

```python
from sklearn.linear_model import LinearRegression

# Create and fit the model
model = LinearRegression() # Create the Linear Regression model object
model.fit(x_values, y_values) # Call the .fit() method, which fits the Linear Regression model to the data provided
```

The Linear Regression model has the slope parameter ($m$), and the intercept parameter ($c$).

We can see what these are as follows:

```python
# Model parameters
print("Slope (m):", model.coef_[0])
print("Intercept (c):", model.intercept_)
```

### 2.1 - Making Predictions

We can use the model to predict `y` values from `x`:

```python
y_pred = model.predict(x_values)

import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))
plt.scatter(x_values, y_values, color="black", label="True data")
plt.plot(x_values, y_pred, color="red", linewidth=2, label="Fitted model")
plt.legend()
plt.title("Linear Regression Fit with scikit-learn")
plt.show()
```

### 2.2 - Evaluating your Model

We can calculate **MSE** and **MAE** to check how well our model fits.

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error

mse = mean_squared_error(y_values, y_pred)
mae = mean_absolute_error(y_values, y_pred)

print("MSE:", mse)
print("MAE:", mae)
```
